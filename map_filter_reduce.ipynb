```python
#functional style tools: map, filter, and reduce

numbers = [1,2,3,4,5,6]
print('Numbers:',numbers)

#map: map() is designed for transformation:
#It takes an input list and transforms each element individually to produce a new list of the same length.
#It cannot look across multiple elements or combine them into a single value.

doubled = list(map(lambda x:x*2, numbers))
print('Doubled:', doubled)

#filter: filter(function, iterable) is designed for selection.
#It allows you to extract elements from a list that meet a specific condition (a test that returns True or False).

evens = list(filter(lambda a:a%2==0, numbers))
print('evens:', evens)

#reduce:reduce() is designed for aggregation:
#It takes an input list and applies a function cumulatively to the items to reduce the entire list down to a single value (like a sum).

from functools import reduce
total = reduce(lambda a,b:a+b, numbers)
print("total is:", total)
```
